Q1. Phone Contact Book
--------------------------------------------------------------
Create a dictionary to store contacts (name -> phone number).
Show a small menu in a while loop:
  1. Add contact
  2. Search contact
  3. Show all contacts
  4. Exit

  - While adding, if the name already exists, print
    "Contact already exists".
  - While searching, if the name is not found, print
    "Contact not found".

```python
contacts ={}
options = ["add contact", "search contact", "show all contacts", "exit"]
user_input = ""

prompt = "Pick an option:\n"
for i, option in enumerate(options):
    prompt+= f"{i+1}. {option}\n"
prompt += "Your choice: "

while True:
    user_input = input(prompt).strip().lower()

    if user_input in [options[0], "1"]:
        name = input("Enter the name of the contact")
        if name in contacts:
            print("Contact already Exists")
        else:
            phone_number = input("Enter the phone number of the contact")
            contacts[name] = phone_number
            print(f"{name}: {phone_number} added to the contact")
            
    elif user_input in [options[1], "2"]:
        name = input("Enter the name to search:")
        if name in contacts:
            print(f"The phone number of {name} is {contacts[name]}")
        else:
            print(f"There is no contact with name {name}")

    elif user_input in [options[2], "3"]:
        if len(contacts) == 0:
            print("The contact book is empty")
        else:
            for name, phone_number in contacts.items():
                print(f"{name}: {phone_number}")
    elif user_input in [options[3], "4"]:
        break
    else:     
        print("Invalid choice, please choose from the options")
```

Q2. Unique Visitors on a Website
--------------------------------------------------------------
A website logged visitor IDs throughout the day (with repeats):
  visits = [101, 205, 101, 330, 205, 412, 101, 330, 509]

Using a set, find and print:
  - The number of total visits
  - The number of unique visitors
  - The unique visitor IDs in sorted order

```python

visits = [101, 205, 101, 330, 205, 412, 101, 330, 509]
unique_visitors = set(visits)

print("The number of total visits:", len(visits))
print("The number of unique visitors:", len(unique_visitors))
print("THe unique vistor IDS in sorted order:", sorted(unique_visitors))
```

Q3. Student Marks Dictionary
--------------------------------------------------------------
Marks of students are stored in a dictionary:
  marks = {"Aman": 82, "Priya": 91, "Rahul": 45,
           "Sneha": 67, "Kiran": 30}

Write a program to:
  - Print each student's name with "Pass" or "Fail"
    (pass mark is 40)
  - Find the topper and their marks
  - Create a new dictionary with only the students who passed

```python
marks = {
    "Aman":82,
    "Priya": 91,
    "Rahul": 45,
    "Sneha": 67,
    "Kiran": 30
}

pass_students = {}

topper=""

for name, mark in marks.items():
    if topper is None or topper == "":
        topper = name
    elif marks[topper] < mark:
        topper = name
    
    if mark>=40:
        print(f"{name}: Pass")
        pass_students[name] = mark
    else:
        print(f"{name}: Fail")
    
print(f"The topper of the class is {topper} with marks {marks[topper]}")
print(f"The students who passed are :\n {pass_students}")
```

Q4. Common Friends
--------------------------------------------------------------
Two friends have these friend lists on a social app:
  riya_friends  = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
  arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

Print:
  - Mutual friends of Riya and Arjun
  - Friends only Riya has
  - All people in either list (without duplicates)
  - Friend suggestions for Riya (Arjun's friends she doesn't know)

```python
riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

print("Mutual friends of Riya and Arjun: ", riya_friends.intersection(arjun_friends))
print("Friends only Riya has: ", riya_friends.difference(arjun_friends))
print("All people in either list: ", riya_friends.symmetric_difference(arjun_friends))
print("Friend suggestions for Riya: ", arjun_friends.difference(riya_friends))
```
