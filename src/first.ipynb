Q1. Movie Ticket Counter
--------------------------------------------------------------
A cinema charges ticket prices based on age:
  - Below 5 years   : Free
  - 5 to 17 years   : Rs. 100
  - 18 to 59 years  : Rs. 200
  - 60 and above    : Rs. 120

Ask the customer for their age and print the ticket price.

Sample run:
  Enter your age: 65
  Your ticket price is Rs. 120

```python
age = int(input("Enter your age:"))
price = "Free"
if age < 5:
    price = "Free"
elif age >= 5 and age < 18:
    price = "Rs 100"
elif age >= 18 and age < 60:
    price = "Rs 200"
else:
    price = "Rs 120"
print(f"Your ticket price is: {price}")
```


Q2. Electricity Bill Calculator
--------------------------------------------------------------
The electricity board charges as follows:
  - First 100 units       : Rs. 3 per unit
  - Next 100 units (101-200): Rs. 5 per unit
  - Above 200 units       : Rs. 8 per unit

Take the number of units consumed and print the total bill.

Sample run:
  Enter units consumed: 250
  Total bill: Rs. 1200
  (100*3 + 100*5 + 50*8 = 300 + 500 + 400)


```python
consumed_units = int(input("Enter the units consumed:"))
first_100_units = 0
next_100_units = 0
above_200_units = 0
total_bill = 0

while consumed_units > 0:
    if consumed_units > 200:
        above_200_units += consumed_units - 200
        total_bill += above_200_units * 8
        consumed_units = 200
    elif consumed_units > 100:
        next_100_units += consumed_units - 100
        total_bill += next_100_units * 5
        consumed_units = 100
    else:
        first_100_units += consumed_units
        total_bill += first_100_units * 3
        consumed_units = 0

print("Total bill:", total_bill)
print(f"({first_100_units}*3 + {next_100_units}*5 + {above_200_units}*8 = {total_bill})")
```

Q4. Savings Piggy Bank
--------------------------------------------------------------
Riya puts money into her piggy bank every day. Her goal is
Rs. 1000. Using a while loop, keep asking how much she saved
today and add it to the total. Stop when the total reaches or
crosses Rs. 1000, then print how many days it took.

Sample run:
  Enter today's savings: 300
  Enter today's savings: 450
  Enter today's savings: 400
  Goal reached in 3 days! Total saved: Rs. 1150

```python
goal = 1000
total = 0
days_counter = 0

while total < goal:
    today_savings = int(input("Enter today's savings: "))
    total += today_savings
    days_counter += 1

print(f"Goal reached in {days_counter} days! Total saved: Rs {total}")
```

Q5. ATM PIN Check
--------------------------------------------------------------
An ATM's correct PIN is 4321. The user gets only 3 attempts.
  - If the PIN is correct, print "Access granted".
  - If wrong, print "Wrong PIN" and ask again.
  - After 3 wrong attempts, print "Card blocked".

Sample run:
  Enter PIN: 1111
  Wrong PIN
  Enter PIN: 4321
  Access granted


```python
correct_pin = "4321"
pin_tries = 3

while pin_tries > 0:
    input_pin = int(input("Enter PIN:"))
    if(input_pin == int(correct_pin)):
        print("Access granted!")
        break
    else:
        pin_tries -= 1
        print(f"PIN incorrect. {pin_tries} tries remaining.")
```

Q6. Even and Odd Counter
--------------------------------------------------------------
A teacher wants to know how many roll numbers from 1 to N are
even and how many are odd. Ask for N and use a for loop to
count them.

Sample run:
  Enter N: 15
  Even numbers: 7
  Odd numbers: 8

```python
even_counter = 0
odd_counter = 0

final = int(input("Enter N:"))

for i in range(1, final + 1):
    if i % 2 == 0:
        even_counter += 1
    else:
        odd_counter += 1

print(f"Even numbers: {even_counter}, Odd numbers: {odd_counter}")
```

Q7. Number Guessing Game
--------------------------------------------------------------
The secret number is 27 (store it in a variable). Keep asking
the player to guess until they get it right. After each wrong
guess, tell them "Too high" or "Too low". At the end, print
how many guesses they took.

Sample run:
  Guess the number: 50
  Too high
  Guess the number: 20
  Too low
  Guess the number: 27
  Correct! You took 3 guesses.


```python
secret_number = 27
guess_counter = 0

while True:
    guess = int(input("Enter your guess: "))
    guess_counter += 1
    if guess == secret_number:
        print(f"Correct! You took {guess_counter} guesses.")
        break
    elif guess < secret_number:
        print("Too low!")
    else:
        print("Too high!")
```

Q8. Exam Result with Grade
--------------------------------------------------------------
Take marks (out of 100) of a student and print the grade:
  - 90 and above : A
  - 75 to 89     : B
  - 50 to 74     : C
  - 35 to 49     : D
  - Below 35     : Fail
If the marks entered are less than 0 or more than 100, print
"Invalid marks".

```python
grade = int(input("Enter grade:"))

if grade >= 90:
    print("A")
elif grade >= 75:
    print("B")
elif grade >= 50:
    print("C")
elif grade >= 35:
    print("D")
elif grade > 100 or grade == 0:
    print("Invalid Marks")
else:
    print("Fail")
```

Q9. Sum of Digits on a Bill Number
--------------------------------------------------------------
A shop gives a free gift if the sum of the digits of the bill
number is more than 20. Take the bill number (as an integer)
and use a while loop with % and // to find the sum of its
digits. Print whether the customer gets a gift.

Sample run:
  Enter bill number: 98765
  Sum of digits: 35
  Congratulations! You win a free gift.

```python
bill_number = int(input("Enter bill number:"))
sum = 0
while bill_number != 0:
    sum += bill_number % 10
    bill_number //= 10

if(sum > 20):
    print("Congratulations! You win a free gift")
else:
    print("Sorry try again.")
```


Q10. Staircase Pattern
--------------------------------------------------------------
A game developer needs a simple staircase made of '*'.
Ask for the number of steps and print the pattern using
nested for loops.

Sample run:
  Enter number of steps: 4
  *
  * *
  * * *
  * * * *

```python
steps = int(input("Enter number of steps:"))

for i in range(steps+1):
    for j in range(i):
        print("*", end=" ")
    print("\n")
```
