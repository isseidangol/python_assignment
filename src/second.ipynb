Q1. Class Marks Report
--------------------------------------------------------------
A teacher has the marks of 8 students stored in a list:
  marks = [78, 45, 92, 33, 67, 88, 51, 29]

Write a program to print:
  - The highest and lowest marks
  - The class average
  - How many students passed (pass mark is 35)

```python
marks = [78, 45, 92, 33, 67, 88, 51, 29]

print("The highest mark is", max(marks), "and the lowest mark is", min(marks))
print("The class average is", sum(marks)/ len(marks))
print("The number of students who passed is:", len([mark for mark in marks if mark>35]))
```

Q2. Grocery Shopping List
--------------------------------------------------------------
Start with an empty shopping list. Keep asking the user to
enter an item name. When the user types "done", stop and
print:
  - The full shopping list, one item per line with a number
  - The total number of items

If the user enters an item that is already in the list, print
"Already added" and do not add it again.

Sample run:
  Enter item: milk
  Enter item: bread
  Enter item: milk
  Already added
  Enter item: done
  1. milk
  2. bread
  Total items: 2

```python
shopping_list =[]

while True:
    list_item = input("Enter an item to add in the shopping list (or type 'done' to finish):")
    if list_item == '':
        print("Please enter a valid item.")
        continue
    if list_item.lower() == 'done':
        break
    if list_item in shopping_list:
        print("Item already exists in the shopping list.")
        continue

    shopping_list.append(list_item)

for index, item in enumerate(shopping_list):
    print(f"{index+1}. {item}")
print("Total items:", len(shopping_list))
```

Q3. Username Validator

A website has these rules for usernames:
  - Length must be between 5 and 12 characters
  - It must not contain any spaces
  - It must not start with a digit

Take a username as input and print "Valid username" or the
reason it is invalid.

  Enter username: 7rahul

Q4. Vowel Counter for a Tweet
--------------------------------------------------------------
A social media app wants to analyse a tweet. Take a sentence
as input and count:
  - Number of vowels
  - Number of consonants
  - Number of spaces

(Ignore digits and special characters for vowel/consonant count.)

```python
tweet = input("Enter your tweet:")
vowel_count = 0
constant_count = 0
blank_spaces = 0

for word in tweet:
    if word.lower() in "aeiou":
        vowel_count += 1
    elif word == " ":
        blank_spaces += 1
    else:
        constant_count += 1

print("Number of vowels:", vowel_count)
print("Number of consonants:", constant_count)
print("Number of blank spaces:", blank_spaces)
```

Q5. Palindrome Name Checker
--------------------------------------------------------------
A game gives bonus points if the player's name reads the same
forwards and backwards (like "Anna" or "Madam").
Take a name as input and check if it is a palindrome.
The check should NOT depend on uppercase/lowercase.

Sample run:
  Enter your name: Anna
  Bonus! Your name is a palindrome.

```python
name = input("Enter your name:")
is_palindrome = True
for i in range(len(name)):
    if name[i] != name[len(name)-1-i]:
        is_palindrome = False
        break

    if i >= (len(name)-1)/2:
        break

if is_palindrome:
    print("Bonus! your name is palindrome")
```

Q6. Temperature Tracker
--------------------------------------------------------------
Temperatures recorded in Hyderabad for a week are:
  temps = [32, 35, 38, 41, 36, 29, 40]
  days  = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

Write a program to:
  - Print the days on which temperature was above 37 (heat alert)
  - Print the hottest day and its temperature
  - Create a new list with temperatures converted to Fahrenheit
    (F = C * 9/5 + 32) and print it

```python
temps = [32,35,38,41,36,29,40] 
days=["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
Fahrenheit = []

for i in range(len(temps)):
    fahrenheit = (temps[i] * 9/5) + 32
    Fahrenheit.append(fahrenheit)
    if temps[i]>37:
        print(f"Heat ALERT:Temperature on {days[i]} is {temps[i]} degrees.")
print("hottest day in the week is", days[temps.index(max(temps))], "with the temperature of", max(temps), "degrees.")
print("Temperatures in Fahrenheit:", Fahrenheit)
```

Q8. Remove Negative Balances
--------------------------------------------------------------
A bank has a list of account balances:
  balances = [1500, -200, 3000, -50, 0, 750, -1000]

Create two new lists using a loop:
  - positive_accounts : balances greater than 0
  - negative_accounts : balances less than 0
Print both lists and the total amount owed by negative accounts.

```python
balances = [1500, -200, 3000, -50, 0, 750, -1000]
positive_accounts = []
negative_accounts = []

for balance in balances:
    if balance >= 0:
        positive_accounts.append(balance)
    else:
        negative_accounts.append(balance)

print("Number of positive accounts:", len(positive_accounts))
print("Number of negative accounts:", len(negative_accounts))
print("total amount owed by negative accounts:", sum(negative_accounts))
```

Q9. Password Strength Checker
--------------------------------------------------------------
Take a password as input. Loop through each character and
check whether it has:
  - At least one uppercase letter
  - At least one lowercase letter
  - At least one digit
  - At least 8 characters

Print "Strong" if all conditions are met, "Medium" if 2 or 3
are met, and "Weak" otherwise.

Hint: use isupper(), islower(), isdigit()

```python
password = input("Create your password:")
conditions_met = 0

if len(password) >= 8:
    conditions_met += 1

if any(c.isupper() for c in password):
    conditions_met += 1

if any(c.islower() for c in password):
    conditions_met += 1

if any(c.isdigit() for c in password):
    conditions_met += 1

if conditions_met == 4:
    print("Password is strong.")
elif conditions_met == 3 or conditions_met == 2:
    print("Password is medium.")
else:
    print("Password is weak.")
```

Q10. Simple Name Initials Generator
--------------------------------------------------------------
An office ID card system needs initials. Take a full name
(e.g., "ruchik kumar sharma") and print:
  - The initials in uppercase with dots  -> R.K.S.
  - The name in title case               -> Ruchik Kumar Sharma
  - The name with only the surname in full -> R. K. Sharma

```python
full_name = input("Enter your full name:")


names = full_name.title().split()
initials = '.'.join([name[0].upper() for name in names]) + "."

name_in_title_case = full_name.title()

name_with_surname = " ".join([name[0].upper() + "." for name in names[:-1]]) + " " + names[-1].upper()




print("Your initials are:", initials)
print("Your name in title case is:", name_in_title_case)
print("Your name with only the surname in full is:", name_with_surname)
```
